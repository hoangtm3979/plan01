```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
dataset = pd.read_csv('/content/drive/MyDrive/churn_modelling.csv')
```

```python
dataset.head()
```

```python
X = dataset.drop(labels=['CustomerId', 'Surname', 'RowNumber', 'Exited'], axis=1)
y = dataset['Exited']
```

```python
X.head()
```

```python
y.head()
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
label1 = LabelEncoder()
X['Geography'] = label1.fit_transform(X['Geography'])
```

```python
label = LabelEncoder()
X['Gender'] = label.fit_transform(X['Gender'])
X.head()
```

```python
X = pd.get_dummies(X, drop_first=True, columns=['Geography'])
X.head()
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0, stratify=y)
```

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
X_train
```

```python
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras import Input

model = Sequential()
model.add(Input(shape=(11,)))
model.add(Dense(11, activation='relu', input_dim=11))
model.add(Dense(128, activation='relu'))
model.add(Dense(1, activation='sigmoid'))

model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense

model = Sequential()
model.add(Dense(11, activation='relu', input_dim=11))
model.add(Dense(128, activation='relu'))
model.add(Dense(1, activation='sigmoid'))

model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

model.fit(X_train, y_train.to_numpy(), batch_size=10, epochs=10, verbose=1)
```

```python
y_pred = model.predict(X_test)
# Convert the predictions to class labels
# (assuming binary classification with a sigmoid activation in the output layer)
y_pred = (y_pred > 0.5).astype("int32")
```

```python
y_pred
```

```python
y_test
```

```python
model.evaluate(X_test, y_test.to_numpy())
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score
```

```python
confusion_matrix(y_test, y_pred)
```

```python
accuracy_score(y_test, y_pred)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(6, 4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=False,
            xticklabels=['Predicted No Exited', 'Predicted Exited'],
            yticklabels=['Actual No Exited', 'Actual Exited'])
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('Confusion Matrix')
plt.show()

accuracy = accuracy_score(y_test, y_pred)
print(f'Accuracy Score: {accuracy:.4f}')
```
