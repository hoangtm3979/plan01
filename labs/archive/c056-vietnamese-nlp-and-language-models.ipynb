```python
# Dùng pipeline như một helper cấp cao (high-level API) trong thư viện Transformers
from transformers import pipeline
```

```python
pipe = pipeline(
    "text-generation",                                # Loại tác vụ: sinh văn bản (text generation)
    model="deepseek-ai/DeepSeek-R1-Distill-Qwen-1.5B",# Tên mô hình pre-trained được lấy từ Hugging Face Hub
    trust_remote_code=True                            # Cho phép chạy code tuỳ chỉnh đi kèm mô hình (cần thiết với một số model đặc biệt)
)
```

```python
# Tạo danh sách các tin nhắn (messages) mô phỏng hội thoại
# - role: "user" nghĩa là nội dung này do người dùng nhập
# - content: nội dung yêu cầu "1+1=?, give a code python to solve this problem"
messages = [
    {"role": "user", "content": "1+1=?, give a code python to solve this problem"},
]
```

```python
a = pipe(messages) # Gọi hàm pipeline đã khởi tạo ở trên với input là "messages".
                    # - Ở đây, pipe là một đối tượng pipeline kiểu "text-generation".
                    # - Tham số "messages" chứa danh sách hội thoại (role, content).
                    # - Kết quả trả về (a) sẽ là đầu ra mô hình sinh văn bản dựa trên nội dung messages.
```

```python
print(a[0]['generated_text'][1]['content'])
# - a là kết quả trả về từ pipe(messages), dạng list.
# - a[0] lấy phần tử đầu tiên trong list (vì thường pipe trả về list 1 phần tử).
# - a[0]['generated_text'] là một list chứa các đoạn hội thoại (cả input và output).
# - [1] chọn đoạn hội thoại thứ 2 trong danh sách (vị trí 1, vì index bắt đầu từ 0).
# - ['content'] lấy phần nội dung (chuỗi văn bản) từ đoạn hội thoại đó.
# → Lệnh này in ra phần trả lời của mô hình sau khi nhận input.
```

Yêu cầu 1: Viết một prompt yêu cầu LLM định nghĩa “học sâu (deep learning)”
dành cho người chưa có nền tảng kỹ thuật.

```python
from transformers import pipeline

# Tạo pipeline với model DeepSeek
pipe = pipeline("text-generation", model="deepseek-ai/DeepSeek-R1-Distill-Qwen-1.5B")

# Prompt yêu cầu chỉ định nghĩa ngắn gọn bằng tiếng Việt
messages = [
    {
        "role": "user",
        "content": "Chỉ trả lời ngắn gọn bằng tiếng Việt: Định nghĩa 'học sâu (deep learning)' dành cho người chưa có nền tảng kỹ thuật. Không giải thích gì thêm."
    }
]

# Gửi prompt cho model
a = pipe(messages, max_new_tokens=100)

# In ra câu trả lời (chỉ lấy phần content của model)
print(a[0]['generated_text'][1]['content'])
```

Yêu cầu 2: Viết prompt yêu cầu LLM liệt kê 5 ứng dụng của AI trong đời sống hằng ngày, trình bày dưới dạng bảng.

```python
from transformers import pipeline
import re

pipe = pipeline("text-generation", model="deepseek-ai/DeepSeek-R1-Distill-Qwen-1.5B", device=0)

messages = [
    {
        "role": "system",
        "content": (
            "Chỉ xuất KẾT QUẢ CUỐI CÙNG. "
            "Không viết suy nghĩ/giải thích/lời dẫn. "
            "Chỉ in BẢNG Markdown với đúng 2 cột: STT | Ứng dụng; đúng 5 dòng dữ liệu. "
            "Bắt đầu NGAY bằng hàng tiêu đề: | STT | Ứng dụng |"
        )
    },
    {
        "role": "user",
        "content": "Liệt kê 5 ứng dụng AI hữu ích trong đời sống hằng ngày. Chỉ ghi tên ứng dụng."
    }
]

a = pipe(
    messages,
    max_new_tokens=80,
    do_sample=False,
    temperature=0.0
)

text = a[0]['generated_text'][1]['content']

# Giữ lại đúng khối bảng đầu tiên
m = re.search(r"(\|\s*STT\s*\|\s*Ứng dụng\s*\|.*?)(?:\n\s*\n|$)", text, flags=re.S)
table = m.group(1) if m else text
print(table)
```
