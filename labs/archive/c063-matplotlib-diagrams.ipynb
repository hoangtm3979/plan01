```python
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch

# Dữ liệu chiến lược
steps = [
    ("1. Thị trường", "Xác định quy mô thị trường, phân khúc, insight khách hàng mục tiêu"),
    ("2. Sản phẩm", "Cải tiến dòng sản phẩm (bền hơn, đa dạng màu/mẫu, dòng cao cấp, thân thiện môi trường)"),
    ("3. Kênh phân phối", "Kết hợp online + offline, kích hoạt đại lý, đồng bộ POSM"),
    ("4. Marketing", "Chiến dịch đánh thức thương hiệu, đánh vào 'nỗi đau', kể câu chuyện HKK 30 năm"),
    ("5. Phản ứng cạnh tranh", "Chuẩn bị gói ưu đãi, tập trung phân khúc ít nhạy cảm về giá")
]

# Vẽ sơ đồ
fig, ax = plt.subplots(figsize=(8, 6))
ax.axis("off")

y = 1.0
for step, detail in steps:
    box = FancyBboxPatch((0.05, y - 0.12), 0.9, 0.1,
                         boxstyle="round,pad=0.02",
                         edgecolor="black", facecolor="#cce5ff")
    ax.add_patch(box)
    ax.text(0.1, y - 0.07, step, fontsize=12, fontweight="bold", va="center")
    ax.text(0.1, y - 0.11, detail, fontsize=10, va="top", wrap=True)
    y -= 0.15

ax.set_xlim(0, 1)
ax.set_ylim(0, 1.05)

plt.tight_layout()
plt.show()
```
