```python
#Bài 1:
# Tạo danh sách để lưu các số thỏa mãn
ket_qua = []

# Duyệt qua tất cả số từ 5000 đến 10000
for so in range(5000, 10001):
    # Kiểm tra điều kiện
    if so % 7 == 0:  # Nếu chia hết cho 7
        if so % 5 != 0:  # Và không chia hết cho 5
            ket_qua.append(str(so))  # Thêm số vào danh sách

# Đáp án
#print(','.join(ket_qua))
output = ",".join(ket_qua)
print("Bài 1 - Kết quả:", output)
```

```python
# Bài 2
# Nhập chuỗi số nhị phân từ người dùng
nhap = input("Nhập các số nhị phân 4 chữ số, cách nhau bằng dấu phẩy: ")

# Tách các số nhị phân thành danh sách
cac_so_nhi_phan = nhap.split(',')

ket_qua = []

for so_nhi_phan in cac_so_nhi_phan:
    # Chuyển từ nhị phân sang thập phân
    so_thap_phan = int(so_nhi_phan, 2)

    # Kiểm tra chia hết cho 5
    if so_thap_phan % 5 == 0:
        ket_qua.append(so_nhi_phan)

# In kết quả
if ket_qua:
    output = ",".join(ket_qua)
    print("Bài 2 - Kết quả số chia hết cho 5:", output)
else:
    print("Không có số nào chia hết cho 5")
```

```python
# Bài 3 (Về nhà): Viết một chương trình chấp nhận chuỗi là các dòng được nhập vào, chuyển các dòng này thành chữ in hoa
# và in ra màn hình. Giả sử đầu vào là: “Hello world”- Thì đầu ra sẽ là: “HELLO WORLD”.


print("Nhập các dòng (Enter 2 lần để kết thúc):")
lines = []
while True:
    line = input()
    if line == "":
        break
    lines.append(line.upper())

print("\nKết quả:")
for line in lines:
    print(line)
```

```python
# Bài 4 (Về nhà): Viết chương trình sắp xếp tuple (name, age, score) theo thứ tự tăng dần, name là string, age và height là
# number. Tuple được đọc từ tệp info.csv. Sắp xếp theo name sau đó sắp xếp theo age, sau đó sắp xếp theo score. Ưu tiên là
# tên > tuổi > điểm. Ví dụ, nếu đầu vào là:
# Tom,19,80
# John,20,90
# Jony,17,91
# Jony,17,93
# Json,21,85
# Thì đầu ra sẽ là: [('John', '20', '90'), ('Jony', '17', '91'), ('Jony', '17', '93'), ('Json', '21', '85'), ('Tom', '19', '80')]

def read_and_sort_tuples(filename):
    # Đọc dữ liệu từ file
    with open(filename, 'r') as file:
        lines = file.readlines()

    # Tách dữ liệu và tạo tuple
    tuples = []
    for line in lines:
        # Loại bỏ ký tự xuống dòng và tách các giá trị
        name, age, score = line.strip().split(',')
        tuples.append((name, age, score))

    # Sắp xếp theo name → age → score
    # Sử dụng lambda để chỉ định key sắp xếp
    sorted_tuples = sorted(tuples, key=lambda x: (x[0], int(x[1]), int(x[2])))

    return sorted_tuples

# Sử dụng hàm
input_filename = 'info.csv'
sorted_data = read_and_sort_tuples(input_filename)
print(sorted_data)
```
