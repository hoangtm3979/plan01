```python
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch

# Create figure
fig, ax = plt.subplots(figsize=(10, 6))
ax.axis('off')

# Title
plt.text(0.5, 1.05, "HKK Market Penetration Strategy\n(Brand Reawakening + Pain-point Attack)",
         ha='center', va='center', fontsize=16, fontweight='bold')

# Boxes data: (text, (x, y))
boxes = [
    ("Current Barrier\n• ISE entrenched\n• Chinese low-price\n• Customer inertia", (0.1, 0.8)),
    ("Customer Pain Points\n• Poor durability (TQ)\n• Limited design/match (ISE)\n• Slow/unstable delivery", (0.5, 0.8)),
    ("HKK Brand Strength\n• 30 years reputation\n• Domestic production\n• Quality control", (0.9, 0.8)),
    ("Strategic Pillar 1:\nBrand Reawakening\n- New brand image\n- Modern storytelling\n- Highlight heritage", (0.1, 0.45)),
    ("Strategic Pillar 2:\nPain-point Attack\n- Guarantee quality\n- Expand designs/colors\n- Fast delivery", (0.5, 0.45)),
    ("Strategic Pillar 3:\nProduct Innovation\n- Anti-rust teeth\n- Seasonal color sets\n- Light & durable", (0.9, 0.45)),
    ("Channel & Dealer Support\n• Risk-free trials\n• Better margins\n• Co-marketing", (0.5, 0.15)),
    ("Outcome Goals\n• Market share growth\n• Stronger North VN presence\n• Dealer loyalty", (0.5, -0.05)),
]

# Function to draw box
def draw_box(text, center):
    x, y = center
    box = FancyBboxPatch((x-0.18, y-0.07), 0.36, 0.14,
                         boxstyle="round,pad=0.02",
                         edgecolor="black", facecolor="#f0f8ff")
    ax.add_patch(box)
    plt.text(x, y, text, ha='center', va='center', fontsize=10)

# Draw all boxes
for text, center in boxes:
    draw_box(text, center)

# Arrows connections
arrowprops = dict(arrowstyle="->", color='black', lw=1.2)
ax.annotate("", xy=(0.5, 0.73), xytext=(0.1, 0.73), arrowprops=arrowprops)
ax.annotate("", xy=(0.5, 0.73), xytext=(0.9, 0.73), arrowprops=arrowprops)
ax.annotate("", xy=(0.1, 0.52), xytext=(0.1, 0.73), arrowprops=arrowprops)
ax.annotate("", xy=(0.5, 0.52), xytext=(0.5, 0.73), arrowprops=arrowprops)
ax.annotate("", xy=(0.9, 0.52), xytext=(0.9, 0.73), arrowprops=arrowprops)
ax.annotate("", xy=(0.5, 0.22), xytext=(0.1, 0.38), arrowprops=arrowprops)
ax.annotate("", xy=(0.5, 0.22), xytext=(0.5, 0.38), arrowprops=arrowprops)
ax.annotate("", xy=(0.5, 0.22), xytext=(0.9, 0.38), arrowprops=arrowprops)
ax.annotate("", xy=(0.5, 0.02), xytext=(0.5, 0.08), arrowprops=arrowprops)

plt.tight_layout()
plt.show()
```
