```python
import torch
from transformers import AutoModel, AutoTokenizer

phobert = AutoModel.from_pretrained("vinai/phobert-base-v2")
tokenizer = AutoTokenizer.from_pretrained("vinai/phobert-base-v2")
```

```python
#@title
!pip install deplacy vncorenlp
!test -d VnCoreNLP || git clone --depth=1 https://github.com/vncorenlp/VnCoreNLP
from vncorenlp import VnCoreNLP
vnp=VnCoreNLP("VnCoreNLP/VnCoreNLP-1.1.1.jar",annotators="wseg")
```

```python

text = "Ông Nguyễn Khắc Chúc  đang làm việc tại Đại học Quốc gia Hà Nội. Bà Lan, vợ ông Chúc, cũng làm việc tại đây."

def word_segmentation(vnp, text):
    word_segment = vnp.tokenize(text)
    result = " ".join(word for sentence in word_segment for word in sentence)
    return result
```

```python
sentence = 'Triệu chứng của cảm cúm'
sen2 = 'Dấu hiệu bị sốt virus'
sentence = word_segmentation(vnp,sentence)
sen2 = word_segmentation(vnp,sen2)

input_ids = torch.tensor([tokenizer.encode(sentence)])
input_ids2 = torch.tensor([tokenizer.encode(sen2)])
vector1 = torch.tensor(input_ids, dtype=torch.float32)
vector2 = torch.tensor(input_ids2, dtype=torch.float32)
print(sentence)
print(sen2)
```

```python
import torch
import torch.nn.functional as F

max_len = max(vector1.size(1), vector2.size(1))
vector1 = F.pad(vector1, (0, max_len - vector1.size(1)))
vector2 = F.pad(vector2, (0, max_len - vector2.size(1)))

# Tính Cosine Similarity
cos_sim = F.cosine_similarity(vector1, vector2)
print("Độ tương đồng Cosine:", cos_sim.item())
```
