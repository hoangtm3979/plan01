## Ý nghĩa hình học / Geometric meaning

Tích vô hướng với **vector đơn vị** cho tọa độ có dấu của phép chiếu.
Ví dụ này minh họa phép chiếu; hướng PC được cho sẵn, chưa được học từ dữ liệu.
Khi áp dụng PCA cho một tập dữ liệu, cần trừ trung bình trước khi chiếu.

A dot product with a unit vector gives a signed projection coordinate.
Here the direction is supplied; a fitted PCA would first center its data.

```python
import numpy as np
u = np.array([1, 0])  # Vector dọc theo trục x
v1 = np.array([1, 0]) # Cùng hướng với u
v2 = np.array([0, 1]) # Vuông góc với u
v3 = np.array([-1, 0]) # Ngược hướng với u

print("u · v1:", np.dot(u, v1))  # Output: 1 (cùng hướng)
print("u · v2:", np.dot(u, v2))  # Output: 0 (vuông góc)
print("u · v3:", np.dot(u, v3))  # Output: -1 (ngược hướng)
```

Vector `[0.707, 0.707]` xấp xỉ hướng 45°. Tôi đối chiếu với vector đơn vị chính xác
để thấy sai số làm tròn. / Compare the rounded direction with an exact unit vector.

```python
# Principal Component (đã chuẩn hóa độ dài = 1)
p = np.array([0.707, 0.707])  # ~ [cos(45°), sin(45°)]

# Điểm dữ liệu
x = np.array([2, 3])

# Chiếu x lên p
projection = np.dot(x, p)
print("Projection:", projection)  # Output: 3.535 (~2*0.707 + 3*0.707)

p_exact = np.array([1.0, 1.0]) / np.sqrt(2.0)
exact_projection = float(np.dot(x, p_exact))
assert np.allclose([np.dot(u, v1), np.dot(u, v2), np.dot(u, v3)], [1, 0, -1])
assert np.isclose(np.linalg.norm(p_exact), 1.0)
assert np.isclose(exact_projection, 5 / np.sqrt(2))
LAB_RESULT = {'rounded_projection': float(projection), 'unit_projection': exact_projection}
print(LAB_RESULT)
```
