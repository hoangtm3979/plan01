# PCA and Vector Projection / PCA và phép chiếu vector

**Bối cảnh / Context:** Tôi luyện tập ý nghĩa hình học của tích vô hướng và phép chiếu lên một hướng thành phần chính (principal component).
**Input:** các vector hai chiều và một vector PC chuẩn hóa gần đúng.
**Output:** dot products và độ lớn của phép chiếu (ví dụ khoảng 3,535).
**Run:** NumPy trong Python; hai cell đã chạy tuần tự thành công trong lần kiểm tra 25/10/2026.




```
# Dot product (tích vô hướng) giữa hai vector không chỉ đơn thuần là một phép tính toán học, mà nó còn có ý nghĩa hình học rất quan trọng trong PCA. Cụ thể:

1. Dot Product đo độ "tương đồng" giữa hai vector
```



````python
import numpy as np
u = np.array([1, 0])  # Vector dọc theo trục x
v1 = np.array([1, 0]) # Cùng hướng với u
v2 = np.array([0, 1]) # Vuông góc với u
v3 = np.array([-1, 0]) # Ngược hướng với u

print("u · v1:", np.dot(u, v1))  # Output: 1 (cùng hướng)
print("u · v2:", np.dot(u, v2))  # Output: 0 (vuông góc)
print("u · v3:", np.dot(u, v3))  # Output: -1 (ngược hướng)
````

→ Giá trị 3.535 cho biết điểm [2, 3] nằm ở đâu dọc theo hướng PC.


````python
# Principal Component (đã chuẩn hóa độ dài = 1)
p = np.array([0.707, 0.707])  # ~ [cos(45°), sin(45°)]

# Điểm dữ liệu
x = np.array([2, 3])

# Chiếu x lên p
projection = np.dot(x, p)
print("Projection:", projection)  # Output: 3.535 (~2*0.707 + 3*0.707)
````
